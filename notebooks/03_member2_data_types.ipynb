## Member 2: Data Types and Formatting Consistency

### Cleaning Decisions & Explanations
1. **Converted `utc_timestamp` to UTC datetime:** Used `pd.to_datetime(..., utc=True)` so all dates and times follow a single, consistent UTC format instead of plain text/object.
2. **Checked and fixed numeric columns:** Verified that all load, wind, solar, and price columns are stored as numbers (`float64`), stripping any stray commas or spaces and converting text-formatted numbers with `pd.to_numeric()`.

```python


# Loading the dataset:
import pandas as pd

raw_df = pd.read_excel("../data/GB_GBN SPECIFIC DATASET.xlsx")
df = raw_df.copy()

print("Dataset loaded successfully! Shape:", df.shape)
```

```python
# Checking- data types BEFORE cleaning
print("--- BEFORE CLEANING ---")
print(df.dtypes)
```

```python
df['utc_timestamp'] = pd.to_datetime(df['utc_timestamp'], utc=True, errors='coerce')

print("Unparseable timestamps (NaT):", df['utc_timestamp'].isna().sum())
print("Timezone:", df['utc_timestamp'].dt.tz)
print("New timestamp dtype:", df['utc_timestamp'].dtype)
```

```python
for col in df.columns:
    if col != 'utc_timestamp' and df[col].dtype == 'object':
        df[col] = df[col].astype(str).str.strip().str.replace(',', '', regex=False)
        df[col] = pd.to_numeric(df[col], errors='coerce')
        print(f"Converted '{col}' to numeric.")

print("Numeric column check complete.")
```

```python
print("--- AFTER CLEANING ---")
print(df.dtypes)
```

### Summary of Member 2 Verification
- All timestamps in `utc_timestamp` are standardized to `datetime64[ns, UTC]` with `0` unparseable values.
- All measurement columns (load, wind, solar, and price) are confirmed as numeric (`float64`).
