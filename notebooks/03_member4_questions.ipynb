```python
import pandas as pd

df = pd.read_csv('cleaned_data.csv')
print(list(df.columns))
```

```python
# Variation of actual electricity demand over the study period

df['utc_timestamp'] = pd.to_datetime(df['utc_timestamp'])
df['year'] = df['utc_timestamp'].dt.year

demand_col = 'GB_GBN_load_actual_entsoe_transparency'

print("Overall Electricity Demand Summary:")
print("Mean:", df[demand_col].mean())
print("Median:", df[demand_col].median())
print("Min:", df[demand_col].min())
print("Max:", df[demand_col].max())
print()

print("Electricity Demand by Year:")
demand_by_year = df.groupby('year')[demand_col].agg(['count', 'mean', 'median', 'min', 'max'])
print(demand_by_year)
```

Q1: How does actual electricity demand vary over the study period?

Actual electricity demand has an average of 34,920.87 MW and a median of 35,306.50 MW, going from a minimum of 511.00 MW in 2019 to a maximum of 57,388.00 MW in 2015. Comparing the years shows that average demand went down steadily, falling from 37,258.11 MW in 2015 to 30,907.01 MW in 2020. (2014 has no valid recorded values because actual demand tracking started in 2015).

```python
# Patterns occurring in wind and solar electricity generation

df['month'] = df['utc_timestamp'].dt.month

wind_col = 'GB_GBN_wind_generation_actual'
solar_col = 'GB_GBN_solar_generation_actual'

print("Wind and Solar Overall Statistics:")
print(df[[wind_col, solar_col]].agg(['count', 'mean', 'median', 'min', 'max']))
print()

print("Average Wind and Solar Generation by Year:")
print(df.groupby('year')[[wind_col, solar_col]].mean())
print()

print("Average Wind and Solar Generation by Month:")
print(df.groupby('month')[[wind_col, solar_col]].mean())
```

Q2: What patterns occur in wind and solar electricity generation?

Average wind generation rose from 3,614.09 MW in 2015 to 5,893.73 MW in 2020, and average solar generation increased from 871.99 MW in 2015 to 1,551.35 MW in 2020. Looking at the monthly averages, wind and solar follow opposite seasonal patterns. Wind generation is highest in the winter months (highest in December at 5,995.53 MW) and lowest in summer (July at 3,149.75 MW). Solar generation is highest in the summer (peaking in May at 1,971.63 MW) and lowest in winter (December at 286.67 MW).

```python
# Wind and Solar Contributions

wind_total = df[wind_col].sum()
solar_total = df[solar_col].sum()
combined_total = wind_total + solar_total

comparison_table = pd.DataFrame({
    'Source': ['Wind', 'Solar'],
    'Total Generation (MW)': [wind_total, solar_total],
    'Average Generation (MW)': [df[wind_col].mean(), df[solar_col].mean()],
    'Share of Combined Total (%)': [
        (wind_total / combined_total) * 100,
        (solar_total / combined_total) * 100
    ]
})

print("Wind vs. Solar Contribution Summary:")
print(comparison_table)
```

Q3: Which renewable energy source contributes more electricity generation?

Wind energy contributes more electricity than solar energy across the study period. Wind generated a total of 227,265,850.00 MW (an average of 4,512.82 MW), making up 79.21% of the combined wind and solar output. In comparison, solar generated a total of 59,653,663.00 MW (an average of 1,184.87 MW), which accounts for 20.79% of the combined generation.
