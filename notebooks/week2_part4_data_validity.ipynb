PART 1:Timestamp Data Type and Validity

```python
#loading the dataset
import pandas as pd
df=pd.read_excel("GB_GBN SPECIFIC DATASET.XLSX")
```

### Dataset Loading

The dataset was loaded into a Pandas DataFrame using `read_excel()`. This provides the data in a tabular format, allowing us to examine its structure, data types and validity. The original dataset is retained unchanged during this investigation.

```python
#checking the utc_timstamp data type
df["utc_timestamp"].dtype
```

### Checking the Timestamp Data Type

The data type of the `utc_timestamp` column was examined to determine whether it was stored in an appropriate date/time format. The result showed that the column initially had an `object` data type, meaning that Pandas was treating the timestamps as text rather than datetime values.

```python
#checking  a few timestamp values
df["utc_timestamp"].head(10)
```

### Examining Timestamp Values

The first ten timestamp values were displayed to inspect their format. The values follow an ISO-style date/time format and end with `Z`, indicating that they are expressed in Coordinated Universal Time (UTC). Although the values contain date and time information, the column is initially stored as an object data type.

```python
#checking for missing values in the utc_timestamp 
df["utc_timestamp"].isna().sum()
```

### Checking for Missing Timestamps

The timestamp column was checked for missing values using `isna().sum()`. The result was zero, meaning that none of the 50,401 records had a missing timestamp. Therefore, all records contain a timestamp value.

```python
#converting from object to datetime format and handling errors by coercing them to NaT (Not a Time)
timestamps=pd.to_datetime(df["utc_timestamp"],errors="coerce",utc=True)
```

### Converting Timestamps to Datetime Format

The timestamp values were converted using `pd.to_datetime()` with UTC enabled. The `errors="coerce"` argument ensures that any values that cannot be converted are represented as `NaT` (Not a Time). The resulting data type was `datetime64[ns, UTC]`, confirming that the conversion was successful and that the timestamps are now recognised as UTC datetime values.

```python
#confirming the conversion and checking the data type of the new timestamps series
timestamps.dtype
```

```python
#checking for missing values in the new timestamps series
timestamps.isna().sum()
```

### Validating the Converted Timestamps

The converted timestamp series was checked for `NaT` values to identify any invalid timestamps that may have resulted from the conversion. The result was zero, indicating that all timestamp values were successfully converted. This confirms that the timestamp column is valid for subsequent time-based analysis.

### Identifying the Timestamps of the Highest Electricity Prices

The timestamps corresponding to the ten highest electricity prices were retrieved to establish when these observations occurred. The results show that several of the highest prices occurred between September and November 2016.

This helps identify periods when unusually high electricity prices were recorded and provides useful information for later analysis of electricity price patterns.

```python
#determining whether the high prices are statistical outliers using the IQR method
Q1 = df["GB_GBN_price_day_ahead"].quantile(0.25)
Q3 = df["GB_GBN_price_day_ahead"].quantile(0.75)

IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR
lower_bound = Q1 - 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Upper Bound:", upper_bound)
print("Lower Bound:", lower_bound)
```

### Applying the Interquartile Range (IQR) Method

The Interquartile Range method was applied to the day-ahead electricity price column to identify statistically unusual observations. The first quartile (Q1) was 33.88, while the third quartile (Q3) was 50.40, giving an IQR of approximately 16.52.

Using the IQR method, the lower boundary was approximately 9.10 and the upper boundary was approximately 75.18. Values outside these boundaries are classified as potential statistical outliers and require further investigation.

```python
#count of values above the upper bound and below the lower bound
outliers = df[
    (df["GB_GBN_price_day_ahead"] < lower_bound) |
    (df["GB_GBN_price_day_ahead"] > upper_bound)
]

print("Number of potential outliers:", len(outliers))
```

### Counting Potential Electricity Price Outliers

The number of electricity price observations falling below the lower IQR boundary or above the upper IQR boundary was calculated. The result identified 1,804 potential outliers.

These observations are statistically unusual according to the IQR method. However, being an outlier does not necessarily mean that a value is incorrect. Electricity prices can fluctuate significantly, so these observations should be investigated before any cleaning decisions are made.

```python
#whether there are negative electricity prices in the GB_GBN_price_day_ahead column
(df["GB_GBN_price_day_ahead"] < 0).sum()
```

# Part 4: Summary of Data Validity and Consistency Checks

The data validity and consistency investigation was conducted to identify potential issues in the Great Britain (GB_GBN) electricity dataset before data cleaning and analysis.

The following findings were established:

1. **Timestamp validity:** The `utc_timestamp` column was initially stored as an object data type. A separate Series was successfully converted into the `datetime64[ns, UTC]` format. No missing or invalid timestamps were identified during the investigation.

2. **Numerical columns:** A total of 15 numerical columns were identified. These cover electricity demand, day-ahead prices, solar energy and wind energy variables.

3. **Solar profile values:** The investigation identified 233 observations greater than 1, including 65 observations greater than 1.1. The maximum value was 1.367, recorded on 11 June 2015 at 12:00 UTC. These values are unusual for a proportion-like profile and require further investigation against the dataset documentation before any cleaning decisions are made.

4. **Unusually high electricity prices:** The highest day-ahead electricity price was 999.00, recorded on 15 September 2016 at 18:00 UTC. The value occurred once, while 799.00 also occurred once. The ten highest prices were examined, and the IQR method identified 1,804 potential statistical outliers.

5. **Negative values:** A total of 81 negative observations were identified in the day-ahead electricity price column. No negative values were found in the other 14 numerical columns. Negative electricity prices should not automatically be treated as errors because they may occur under genuine electricity market conditions.

6. **Zero values:** Zero observations were found in several numerical columns, particularly solar generation (20,037) and solar profile (17,655). These values may represent periods when solar energy was not being generated and should not automatically be removed.

7. **Missing values:** Missing observations were identified across all 15 numerical columns. The highest missing counts were found in the renewable energy capacity and profile variables, indicating that these columns require particular attention during data cleaning.

### Overall Conclusion

The investigation identified several data-quality issues that require attention, including unusual solar profile values, extreme electricity prices, negative prices, zero observations and missing values. However, these observations should not automatically be classified as errors because some may represent genuine electricity market and renewable energy conditions.
