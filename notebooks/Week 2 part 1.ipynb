## 1. Import Pandas and load the dataset (relative path)
The Excel file `GB_GBN SPECIFIC DATASET.xlsx` is stored in the project's `data/raw/` folder. The code uses relative paths only, so it works from the group GitHub repository. The file is then read with `pd.read_excel()`.

```python
# Dataset loading and initial inspection 
import pandas as pd
from pathlib import Path

FILE_NAME = "GB_GBN SPECIFIC DATASET.xlsx"

candidates = [
    Path("..") / "data" / "raw" / FILE_NAME,   # notebooks/ -> data/raw/
    Path("data") / "raw" / FILE_NAME,          # notebook next to data/
    Path(FILE_NAME),                           # file in same folder
]
DATA_PATH = next((p for p in candidates if p.exists()), None)

print("Using:", DATA_PATH)
print("File found:", DATA_PATH is not None)

df = pd.read_excel(DATA_PATH)
df["utc_timestamp"] = pd.to_datetime(df["utc_timestamp"], utc=True)
```

```python
print(df.head(10))
```

**The df.head(10) shows:** Each row is one hour(`utc_timestamp`), the first date was 2014-12-31 23:00 UTC. The very first row has no load, price or generation values (NaN). Moreover the early rows have no price. The solar and wind (On and off shore) capacity columns are filled from the first row.

```python
print(df.tail(5))
```

**The df.tail(5) shows:** The data ends at 2020-09-30 23:00 UTC, Furthermore it shows that each row is still one row per hour. The capacity columns for both wind and solar capacity are empty (NaN) at the end of the file, indicating they stop before the dataset does.

## 8. General structure of the dataset

- **Size:** 50,401 rows and 16 columns.
- **What a row is:** one hour of electricity data for Great Britain, from 2014-12-31 23:00 to 2020-09-30 23:00 UTC (about 5.75 years).
- **Column groups:** one time column (`utc_timestamp`); demand (`load_actual`, `load_forecast`); price (`price_day_ahead`); and solar and wind groups, each with capacity, actual generation and profile (capacity factor). Wind is split into total, offshore and onshore.
- **Data types:** 1 datetime column and 15 numeric (`float64`) columns.
- **Completeness:** load, price and generation columns are almost complete. The capacity and profile columns are missing about 13% of rows (they end around the end of 2019).
- **Things to investigate in the next Week 2 sections:** the missing values, the very low load minimum (511 MW), the price maximum (999.0) and solar profile values above 1.
