```python
import pandas as pd

df = pd.read_excel('GB_GBN SPECIFIC DATASET.xlsx')
gb_cols = [col for col in df.columns if col.startswith('GB_GBN')]
df_filtered = df[gb_cols]

# validate_columns() 
def validate_columns(dataframe, required_cols):
    missing_cols = []
    for col in required_cols:
        if col not in dataframe.columns:
            missing_cols.append(col)
            
    if len(missing_cols) == 0:
        print("Valid. Necessary columns are present.")
    else:
        print("Invalid. Missing columns:", missing_cols)

# summarize_missingness() 
def summarize_missingness(dataframe):
    total_rows = len(dataframe)
    missing_counts = dataframe.isnull().sum()
    missing_percentages = (missing_counts / total_rows) * 100
    
    summary = pd.DataFrame({
        'Missing Count': missing_counts,
        'Missing Percentage (%)': missing_percentages
    })
    return summary

print("Testing validate_columns()")
validate_columns(df_filtered, gb_cols)
print()

print("Testing summarize_missingness()")
missing_report = summarize_missingness(df_filtered)
print(missing_report)
```
