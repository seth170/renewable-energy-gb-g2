```python
import pandas as pd

# 1. Load the dataset 
df = pd.read_excel('GB_GBN SPECIFIC DATASET.xlsx')

# Filter columns that start with GB_GBN as assigned to your group
meaningful_cols = [col for col in df.columns if col.startswith('GB_GBN')]
df_filtered = df[meaningful_cols]

# 2. df.shape 
print("Dataset Shape")
print(df_filtered.shape)
print()

# 3. df.columns
print("Dataset Columns")
print(df_filtered.columns)
print()

# 4. df.head()
print("First 5 Rows")
print(df_filtered.head())
print()

# 5. Proving at least 15,000 rows and 10 meaningful columns
rows = df_filtered.shape[0]
columns = df_filtered.shape[1]

print("Validation Check")
if rows >= 15000 and columns >= 10:
    print("The dataset meets the requirement of at least 15,000 rows and 10 columns.")
else:
    print("The dataset does not meet the requirements.")
```
