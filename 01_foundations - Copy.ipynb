# Investigation Foundations: Python Basics

This notebook covers fundamental Python concepts tailored to an investigative workflow: variables, lists, dictionaries, loops, conditionals, and domain-specific helper functions.

```python
# 1. Variables
case_id = "INVESTIGATION-2026-04"
suspect_name = "Alex Vance"
target_amount = 150000.00
is_active_case = True

print(f"Case ID: {case_id}")
print(f"Suspect: {suspect_name}")
print(f"Target Amount: ${target_amount:,.2f}")
print(f"Active Case Status: {is_active_case}")
```

```python
# 2. List
evidence_list = [
    "Bank Transfer Log #1042",
    "Encrypted Chat Backup",
    "Hotel Receipt - Downtown",
    "Flight Itinerary #AA-882",
    "Flash Drive (16GB)"
]

print(f"Total evidence items collected: {len(evidence_list)}")
print(f"First evidence item: {evidence_list[0]}")
```

```python
# 3. Dictionary
subject_profile = {
    "name": "Alex Vance",
    "age": 34,
    "role": "Financial Analyst",
    "associated_accounts": ["ACCT-8891", "ACCT-4420", "ACCT-9011"],
    "flagged": True,
    "total_unexplained_deposits": 48500.00
}

print("Subject Profile:")
for key, value in subject_profile.items():
    print(f"  {key}: {value}")
```

```python
# 4. Loop
print("Processing Evidence Files:")
for index, item in enumerate(evidence_list, start=1):
    print(f"  Item {index}: Scanning {item}...")
```

```python
# 5. Conditional
transaction_amount = 55000.00
alert_threshold = 10000.00

if transaction_amount > alert_threshold:
    print(f"ALERT: Transaction of ${transaction_amount:,.2f} exceeds threshold (${alert_threshold:,.2f}). Escalating for review.")
else:
    print(f"Transaction of ${transaction_amount:,.2f} is within normal parameters.")
```

```python
# 6. Two small functions relevant to the investigation

def flag_suspicious_account(account_id, balance, limit=25000.0):
    """Checks if an account balance exceeds a suspicious threshold."""
    if balance > limit:
        return f"[FLAGGED] Account {account_id} exceeds threshold with balance: ${balance:,.2f}"
    return f"[CLEAR] Account {account_id} is within safe limit (${balance:,.2f})"

def format_case_summary(case_id, suspect, total_evidence_count):
    """Formats a quick brief summary of the investigation case."""
    summary = (
        f"=== CASE BRIEF ===\n"
        f"Case: {case_id}\n"
        f"Primary Subject: {suspect}\n"
        f"Evidence Items Logged: {total_evidence_count}\n"
        f"=================="
    )
    return summary

# Testing the functions
print(flag_suspicious_account("ACCT-8891", 48500.00))
print(flag_suspicious_account("ACCT-4420", 12000.00))
print()
print(format_case_summary(case_id, suspect_name, len(evidence_list)))
```
